```python
import os 
import sys
import numpy as np
thincurr_python_path = '/home/clair/repos/install_release'
if thincurr_python_path is not None:
    sys.path.append(os.path.join(thincurr_python_path,'python'))
from OpenFUSIONToolkit.io import write_oft_xml
from OpenFUSIONToolkit.ThinCurr.coils import ThinCurr_Icoil, ThinCurr_Vcoil, ThinCurr_XML
```

```python
xml = ThinCurr_XML()

xml.set_eta([5.0e-6, 4e20])

vcoil1 = ThinCurr_Vcoil(
    name="VCoil_1",
    resistivity_per_length=1.7e-8,
    radius=1.0
)
for sub_coils in [[1.5, 0.0], [1.5, 0.2]]:
    vcoil1.add_subcoil(
        RZ=np.r_[sub_coils[0], sub_coils[1]],
        scale=1.0
    )
xml.add_Vcoil(vcoil1)

# Write XML file
write_oft_xml([xml], "valen_test.xml")
print("Created valen_test.xml")
```
